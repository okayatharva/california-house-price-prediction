```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load the raw dataset using relative path from notebooks/
df = pd.read_csv("../data/raw/california_housing.csv")

# 1. Dataset Inspection
print("--- Dataset Info ---")
print(df.info())

print("\n--- Summary Statistics ---")
print(df.describe())

print("\n--- Missing Values Per Column ---")
print(df.isnull().sum())

# 2. Target Distribution Plot
plt.figure(figsize=(6, 4))
sns.histplot(df["MedHouseVal"], bins=40, kde=True)
plt.title("Distribution of Median House Value")
plt.xlabel("Median House Value ($100,000s)")
plt.savefig("housing_dist.png")
plt.show()

# 3. Correlation Heatmap
plt.figure(figsize=(8, 6))
sns.heatmap(df.corr(), annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Feature Correlation Heatmap")
plt.savefig("housing_corr.png")
plt.show()
```
